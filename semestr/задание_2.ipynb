# A - Задача на вычисление

Система навигации самолёта информирует пассажира о том, что полёт проходит на высоте 37 000 футов. Выразите высоту полёта в метрах. Считайте, что 1 фут равен 30,5см.

## Задание

Напишите программу на языке Python для решения задачи с использованием переменных.

```python
# Ваш код здесь
# 1. Входные данные. Здесь нужно определить переменные для хранения входных данных и присвоить им значения.
height_feet = 37000
feet_to_cm = 30.5
# 2. Обработка данных. Здесь нужно вычислить искомое значение и сохранить его в переменной.
height_cm = height_feet * feet_to_cm
height_meters = height_cm / 100
# 3. Вывод результатов. Здесь нужно вывести вычисленное значение на экран.
print(f"Высота полёта: {height_meters} метров")
```

# B - Задача на ввод и вывод

В летнем лагере на каждого участника полагается Xг сахара в день. В лагере Y человек. Сколько килограммовых упаковок сахара понадобится на весь лагерь на 5 дней?

## Задание

Напишите программу на языке Python для решения задачи с использованием переменных.

```python
# Ваш код здесь
# 1. Входные данные. Здесь нужно определить переменные для хранения входных данных и ввести значения с клавиатуры.
X = int(input("Введите количество грамм сахара на одного человека: "))
Y = int(input("Введите количество человек в лагере: "))
# 2. Обработка данных. Здесь нужно вычислить искомое значение и сохранить его в переменной.
total_grams = X * Y * 5
total_kg = total_grams / 1000
packages = total_kg
# 3. Вывод результатов. Здесь нужно вывести вычисленное значение на экран.
rint(f"Понадобится {packages} килограммовых упаковок сахара.")
```

# C - Задача про Елисея

Елисей меняет прическу исходя из времени года. Зимой он отращивает волосы, весной он носит челку на сторону, летом он стрижется налысо, осенью носит короткую стрижку.

## Задание

Напишите программу на языке Python, которая по введенному времени года выведет прическу для Елисея.

```python
# Ваш код здесь
# 1. Входные данные
season = input("Введите время года (зима/весна/лето/осень): ").lower().strip()

# 2. Обработка данных
if season == "зима":
    hairstyle = "отращивает волосы"
elif season == "весна":
    hairstyle = "носит челку на сторону"
elif season == "лето":
    hairstyle = "стрижется налысо"
elif season == "осень":
    hairstyle = "носит короткую стрижку"
else:
    hairstyle = "неизвестная прическа (время года указано неверно)"

# 3. Вывод результата
print(f"Елисей {hairstyle}.")
```

# D - Задача на списки

Нужно сформировать список  lst из 20 элементов случайным образом. Элементы списка должны быть в диапазоне [-50;50]. Выведите элементы списка на экран.

```python
# Ваш код здесь
import random
lst = [random.randint(-50, 50) for _ in range(20)]
print("Сформированный список:", lst)
```

#### Задание 1: Найдите длину списка

```python
# Ваш код здесь
print("Длина списка:", len(lst))
```

#### Задание 2: Найдите максимальный и минимальный элемент списка

```python
# Ваш код здесь
print("Максимальный элемент:", max(lst))
print("Минимальный элемент:", min(lst))
```

#### Задание 3: Выведите список в обратном порядке

```python
# Ваш код здесь
 print("Список в обратном порядке:", lst[::-1])
```

#### Задание 4: Создайте копию списка в переменной mlst и отсортируйте список mlst по возрастанию

```python
# Ваш код здесь
mlst = lst.copy()
mlst.sort()
print("Отсортированный список:", mlst)
```

#### Задание 5: Дано число K=4: Выведите первые K элементов списка

```python
# Ваш код здесь
K = 4
print(f"Первые {K} элемента:", lst[:K])
```

#### Задание 6: Выведите элементы списка, которые заканчиваются на 0

```python
# Ваш код здесь
ends_with_zero = [x for x in lst if x % 10 == 0]
print("Элементы, заканчивающиеся на 0:", ends_with_zero)
```

#### Задание 7: Найдите произведение элементов списка, которые меньше среднего арифметического

```python
# Ваш код здесь
average = sum(lst) / len(lst)
product = 1
for num in lst:
    if num < average:
        product *= num

print("Среднее арифметическое:", round(average, 2))
print("Произведение элементов меньше среднего:", product)
```
